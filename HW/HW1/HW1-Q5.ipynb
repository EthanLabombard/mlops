```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats
```

# a)

```python
n = 1000000
sh_analysis = stats.expon.rvs(scale = 10, size = n)
data_ext = stats.norm.rvs(15,3,n)
dev_exp = stats.norm.rvs(3,.5,n)
dev_mod = stats.poisson.rvs(mu = 30, size = n)
process_int = stats.norm.rvs(50,20,n)
data_pipe = stats.poisson.rvs(mu = 100, size = n)
model_prod = stats.norm.rvs(30,5,n)
```

# b)

## i)

### 1)

```python
tot_days = sh_analysis + data_ext + dev_exp + dev_mod + process_int + data_pipe + model_prod
print("Mean:", np.mean(tot_days))
print("Median:", np.median(tot_days))
print("95% Confidence Interval:", np.quantile(tot_days, .95))
```

### 2)

```python
print("SH Analysis Overall Time:", np.mean(sh_analysis))
print("Data Extract Overall Time:", np.mean(data_ext))
print("Dev Exploration Overall Time:", np.mean(dev_exp))
print("Dev Modeling Overall Time:", np.mean(dev_mod))
print("Process Integration Overall Time:", np.mean(process_int))
print("Data Pipeline Overall Time:", np.mean(data_pipe))
print("Model Productionization Overall Time:", np.mean(model_prod))
```

```python
print("SH Analysis Uncertainty in Overall Time:", np.std(sh_analysis))
print("Data Extract Uncertainty in Overall Time:", np.std(data_ext))
print("Dev Exploration Uncertainty in Overall Time:", np.std(dev_exp))
print("Dev Modeling Uncertainty in Overall Time:", np.std(dev_mod))
print("Process Integration Uncertainty in Overall Time:", np.std(process_int))
print("Data Pipeline Uncertainty in Overall Time:", np.std(data_pipe))
print("Model Productionization Uncertainty in Overall Time:", np.std(model_prod))
```

The step that contributes most to the overall time is the data pipeline step because this step has the higest mean time of all the steps. The step that contributes most to the uncertainty of the overall time is the process integration step because this step has the highest standard deviation in time of all the steps. 

## ii)

```python
data_pipe_NEW = stats.poisson.rvs(mu = 150, size = n)
tot_days_NEW = sh_analysis + data_ext + dev_exp + dev_mod + process_int + data_pipe_NEW + model_prod
delay = tot_days_NEW - tot_days
print("Delay Mean:", np.mean(delay))
print("Delay 95% Confidence Interval:", np.quantile(delay, .95))
```
