```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
import numpy as np
import pandas as pd
```

```python
data = pd.read_csv(r"..\_data\sampregdata.csv")

model = LinearRegression()
X = data[['x4']]
y = data['y']
model.fit(X,y)

ypred = model.predict(X)
r2_score(y, ypred)
```
