```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
```

```python
mn = [30, 1000] # price and then volume
```

```python
covar = [
    [100, -700],
    [-700, 10000]
]
```

```python
mvn = stats.multivariate_normal(mean=mn, cov=covar)
```

```python
samp = mvn.rvs(size=200)
```

```python
plt.scatter(samp[:, 1], samp[:, 0])
plt.xlabel('Volume')
plt.ylabel('Price')
plt.show()
```

```python
samp = np.column_stack((samp, samp[:, 0] * samp[:, 1]))
```

```python
print(np.mean(samp[:,0]),np.mean(samp[:,1]), np.mean(samp[:,0])*np.mean(samp[:,1]),np.mean(samp[:,2]))
```
